`router-mode = length`: estimated prompt tokens (`chars/4`) select the tier.
`< threshold` → **pym** (small) · `< 4×threshold` → **stark** · otherwise **thanos**.

```python
import sys; sys.path.insert(0, '.')
from heimdal_helpers import *

set_router(router_mode='length', length_threshold_tokens='200')
import time; time.sleep(60)  # named value propagation
```

```python
import sys; sys.path.insert(0, '.')
from heimdal_helpers import *

filler = "The Avengers assembled in the tower and discussed the next mission in great detail. "
rows = []
for label, n in [('short', 1), ('medium', 40), ('long', 200)]:
    p = f"Summarise this in one sentence: {filler * n}"
    rows.append({'prompt': f'{label} (~{len(p)//4} tok)', **ask(p, user='loki')})
show(rows)
```
