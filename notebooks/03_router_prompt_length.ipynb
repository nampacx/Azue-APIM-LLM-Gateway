# 03 · Model router – by prompt length
`router-mode = length`: estimated prompt tokens (`chars/4`) select the tier.
`< threshold` → **pym** (small) · `< 4×threshold` → **stark** · otherwise **thanos**.

```python
import sys; sys.path.insert(0, '.')
from heimdal_helpers import *

set_router(router_mode='length', length_threshold_tokens='200', wait_s=60)
```

```python
import sys; sys.path.insert(0, '.')
from heimdal_helpers import *

FILLER = "The Avengers assembled in the tower and discussed the next mission in great detail. "
SIZES = {'short': 1, 'medium': 20, 'long': 200}   # how often FILLER is repeated
prompts = {f'{label} (~{len(FILLER) * n // 4} tok)': f"Summarise this in one sentence: {FILLER * n}" for label, n in SIZES.items()}
show(ask_all(prompts, user='loki'))
```
