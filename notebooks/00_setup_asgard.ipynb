## 🚀 Deploy (once, from a terminal in the repo root)
`azd up` is the main path. It deploys everything and writes the `.env` this notebook reads:
```
pip install -r requirements.txt
azd auth login && az login
azd env new heimdal-dev
azd env set AZURE_LOCATION swedencentral
azd env set HEIMDAL_PUBLISHER_EMAIL you@contoso.com
azd up
```
Alternative without azd (same Bicep, same `.env`):
```
python scripts/deploy.py --check-models
python scripts/deploy.py --email you@contoso.com
```
Optional, before deploying: `HEIMDAL_ENABLE_MAI=true` (needs quota) and `HEIMDAL_ENABLE_OLLAMA=true` (local model, notebook 09). Details: the README, section *Deploy with `azd`*.
