# 06 · S.H.I.E.L.D. tools via MCP
An **MCP server on Azure Functions (Python)** exposed through APIM as an MCP server: subscription auth, rate limiting, and the function key injected by the gateway – clients never see the backend.

```python
import sys; sys.path.insert(0, '.')
from heimdal_helpers import *

import asyncio, json
from mcp import ClientSession
from mcp.client.streamable_http import streamablehttp_client

async def main():
    headers = {'api-key': user_key('thor')}
    async with streamablehttp_client(env('HEIMDAL_MCP_URL'), headers=headers) as (r, w, _):
        async with ClientSession(r, w) as s:
            await s.initialize()
            tools = await s.list_tools()
            print([t.name for t in tools.tools])
            res = await s.call_tool('locate_infinity_stone', {'name': 'time'})
            print(res.content[0].text)
            res = await s.call_tool('get_mission_status', {'name': 'bifrost-repair'})
            print(res.content[0].text)
await main()
```
